```python
import pandas as pd

# Load the Excel files
finance = pd.read_excel('Team Two Igbo-Finance-Dataset.xlsx')
agriculture = pd.read_excel('igbo_Agric_translation_dataset_cleaned.xlsx')
health = pd.read_excel('Copy_of_CLEANED_IGBO_HEALTH.xlsx')

# Combine all datasets into one
combined = pd.concat([finance, agriculture, health], ignore_index=True)

# Shuffle to mix up the order
combined = combined.sample(frac=1, random_state=42).reset_index(drop=True)

# Save the merged dataset
combined.to_excel('combined_english_igbo_dataset.xlsx', index=False)

print("✅ Combined dataset created successfully!")
print("Total entries:", len(combined))
```
